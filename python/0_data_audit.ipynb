```python
import pandas as pd
from config import RAW_SALES_PATH, PRODUCTS_CLEAN_PATH

df = pd.read_excel(RAW_SALES_PATH)

print("Shape:", df.shape)
print("\nColumns:\n", df.columns.tolist())
print("\nDtypes:\n", df.dtypes)
print("\nSample row:\n", df.iloc[0])
print("\nNull counts:\n", df.isnull().sum())
```

```python
df[['user_id', 'user_name', 'review_title']].iloc[[0]]
```

```python
df['n_reviewers'] = df['user_id'].astype(str).str.split(',').apply(len)
print(df['n_reviewers'].describe())
```

```python
print(df[['actual_price','discounted_price','discount_percentage','rating','rating_count']].describe())
```

```python
# Print the exact count of hidden non-numeric placeholder rows
print("Mismatched or missing rating values:", pd.to_numeric(df['rating'], errors='coerce').isnull().sum())
```

```python
df = df[df['rating'] != '|'].copy()
df['rating'] = df['rating'].astype(float)
print("Rows after dropping bad rating:", df.shape[0])
```

```python
df.to_csv(PRODUCTS_CLEAN_PATH, index=False)
print("Saved cleaned file to products_clean.csv")
```
